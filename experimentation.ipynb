```python
def split_dataset(avg_dataset):
    X = np.array(avg_dataset.values)
    antigen = list(avg_dataset.index.get_level_values('Peptide'))
    y = np.array(list(map(lambda x: ANTIGENS.index(x), antigen)))
    y = to_categorical(y)
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=TEST_PROPORTION, random_state=42)
    return X_train, X_test, y_train, y_test
```

```python
def print_reconstruction(trained_model):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    # Plot the accuracy on the first subplot
    ax1.plot(trained_model.history['mse'], label='Accuracy')
    ax1.plot(trained_model.history['val_categorical_accuracy'], label='Validation Accuracy')
    ax1.set_title('Model Accuracy')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Accuracy')
    ax1.legend()

    # Plot the loss on the second subplot
    ax2.plot(trained_model.history['loss'], label='Loss')
    ax2.plot(trained_model.history['val_loss'], label='Validation Loss')
    ax2.set_title('Model Loss')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Loss')
    ax2.legend()

    # Adjust the spacing between subplots
    plt.tight_layout()

def do_training(X_train, X_test, model, batch_size):
    trained_model = model.fit(X_train, X_train,
        epochs=1000,
        shuffle=True,
        validation_data=(X_test, X_test),
        #callbacks=[training_callback],
        batch_size=batch_size,
        verbose=0)
    print_reconstruction(trained_model)
    return trained_model
    
    
```
